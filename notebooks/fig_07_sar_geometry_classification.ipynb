# Geometry-aligned classification of unfocused SAR phase histories

We ask whether the query-induced circuit bias can be tied to a physical acquisition. In a controlled synthetic-aperture-radar (SAR) experiment, the carrier frequency, range-frequency and azimuth-wavenumber sampling, squint and reference slant range fix a scalar query $f^\star_{\rm geom}$ before any scene or label is drawn. The input is a small complex phase-history patch in azimuth- and range-wavenumber coordinates, upstream of focusing, whose nonseparable phase follows the $\omega$-$k$ dispersion surface. The circuit applies the leading mixed term of this geometry with $n^2=9$ cross-register $R_{ZZ}$ gates on two three-qubit registers, traces out the range register and classifies the target's azimuth cell with a shallow trainable readout.

All models receive the same amplitude-encoded phase history and the same readout and differ only in the fixed query: geometry aligned ($f=f^\star_{\rm geom}$), mismatched ($f=f^\star_{\rm geom}/2$) or no query ($f=0$). The figure tests (a) whether test accuracy peaks at the geometry-derived query, (b) whether a training-only optimum $\widehat f$ tracks the geometry across reference ranges, and (c) whether alignment improves sample efficiency. Panel (b) is a diagnostic; the aligned classifier always takes its query from the geometry and never selects it from data.

**Outputs.** `figures/fig_07_sar_geometry_classification.pdf` (and `.png`) — the three-panel figure; `figures/fig_07_sar_geometry_classification_geometry.pdf` (and `.png`) — exact versus compact geometry interaction; `results/sar_geometry/` — raw per-fit results, run configuration and aggregated summaries.

## Acquisition geometry and fixed query

For the sampled range frequencies and azimuth wavenumbers,

$$
k_{r,d}=\frac{4\pi}{c}\left[f_c+\left(d-\frac{Q-1}{2}\right)\Delta f_r\right],
\qquad
k_{x,a}=k_{x,0}+\left(a-\frac{Q-1}{2}\right)\Delta k_x .
$$

The $\omega$-$k$ surface is $k_{y,ad}=\sqrt{k_{r,d}^2-k_{x,a}^2}$. Terms that depend only on $a$ or only on $d$ are separable single-register phases; removing them leaves the interaction

$$
\widetilde k_{y,ad}=k_{y,ad}-\overline k_{y,a\cdot}-\overline k_{y,\cdot d}+\overline k_y .
$$

Around the patch center, with centered indices $\widetilde a$ and $\widetilde d$,

$$
\widetilde k_{y,ad}\simeq\kappa_{xr}\,\Delta k_x\Delta k_r\,\widetilde a\widetilde d,
\qquad
\kappa_{xr}=\left.\frac{\partial^2 k_y}{\partial k_x\partial k_r}\right|_0
=\frac{k_{x,0}k_{r,0}}{(k_{r,0}^2-k_{x,0}^2)^{3/2}} .
$$

A target at reference slant range $R_0$ therefore carries the leading nonseparable phase

$$
f^\star_{\rm geom}\,\widetilde a\widetilde d,
\qquad
f^\star_{\rm geom}=R_0\,\kappa_{xr}\,\Delta k_x\Delta k_r .
$$

All of these quantities are fixed before any scene or label is drawn, so $f^\star_{\rm geom}$ is a single query for the whole task rather than sample metadata. The circuit uses only the compact generator $\widetilde a\widetilde d$, whereas the scenes are generated from the full sampled interaction $H^{\rm exact}_{ad}=\widetilde k_{y,ad}/(\kappa_{xr}\Delta k_x\Delta k_r)$. A successful aligned model can therefore not be attributed to constructing the data with the exact inverse of the circuit.

We sample the patch, form the interaction and derive $\kappa_{xr}$ and $f^\star_{\rm geom}$. The three arms use $f^\star_{\rm geom}$, $f^\star_{\rm geom}/2$ and $0$.

```python
TWO_PI = 2 * np.pi
INDEX = np.arange(Q, dtype=float)
CENTERED = INDEX - (Q - 1) / 2
CELL_WIDTH = TWO_PI / Q

KR_CENTER = 4 * np.pi * CARRIER_HZ / C0
DELTA_KR = 4 * np.pi * RANGE_FREQUENCY_STEP_HZ / C0
KR = KR_CENTER + DELTA_KR * CENTERED
KX = AZIMUTH_WAVENUMBER_CENTER + AZIMUTH_WAVENUMBER_STEP * CENTERED

if not np.all(KR[None, :] ** 2 > KX[:, None] ** 2):
    raise ValueError("evanescent samples: require k_r^2 > k_x^2 over the patch")

KY = np.sqrt(KR[None, :] ** 2 - KX[:, None] ** 2)
KY_INTERACTION = (
    KY
    - KY.mean(axis=1, keepdims=True)
    - KY.mean(axis=0, keepdims=True)
    + KY.mean()
)

KAPPA_XR = (
    AZIMUTH_WAVENUMBER_CENTER
    * KR_CENTER
    / (KR_CENTER**2 - AZIMUTH_WAVENUMBER_CENTER**2) ** 1.5
)
MIXED_SCALE = KAPPA_XR * AZIMUTH_WAVENUMBER_STEP * DELTA_KR

# exact dimensionless interaction and its compact leading approximation
H_EXACT = KY_INTERACTION / MIXED_SCALE
H_COMPACT = np.outer(CENTERED, CENTERED)

F_STAR = REFERENCE_RANGE_M * MIXED_SCALE
F_WRONG = 0.5 * F_STAR
ARMS = {
    "aligned": F_STAR,
    "wrong": F_WRONG,
    "none": 0.0,
}
```

The query scan covers $f/f^\star_{\rm geom}\in[0,2]$ and always contains the three arm values. For geometry tracking we vary the reference range $R_0$ and scan a common grid of absolute queries around the corresponding values of $f^\star_{\rm geom}$.

```python
GLOBAL_SCALES = np.unique(
    np.concatenate(
        [np.linspace(0.0, 2.0, P["global_points"]), [0.0, 0.5, 1.0]]
    )
)
GLOBAL_F = F_STAR * GLOBAL_SCALES

RECOVERY_RANGES = np.asarray(P["recovery_ranges_m"], dtype=float)
RECOVERY_F_STAR = RECOVERY_RANGES * MIXED_SCALE
RECOVERY_F_GRID = np.linspace(
    0.75 * RECOVERY_F_STAR.min(),
    1.25 * RECOVERY_F_STAR.max(),
    P["recovery_f_points"],
)
```

We quantify how well the compact generator describes the sampled interaction: the least-squares coefficient of $\widetilde a\widetilde d$ in $H^{\rm exact}$, the fraction of variance it explains ($R^2$), and the largest residual phase $f^\star_{\rm geom}\,|H^{\rm exact}-\widetilde a\widetilde d|$ at the aligned query.

```python
bilinear_coefficient = float(
    np.sum(H_EXACT * H_COMPACT) / np.sum(H_COMPACT**2)
)
bilinear_r2 = float(
    1.0
    - np.sum((H_EXACT - bilinear_coefficient * H_COMPACT) ** 2)
    / np.sum((H_EXACT - H_EXACT.mean()) ** 2)
)
maximum_aligned_residual = float(
    np.max(np.abs(F_STAR * (H_EXACT - H_COMPACT)))
)

GEOMETRY = {
    "carrier_hz": CARRIER_HZ,
    "range_frequency_step_hz": RANGE_FREQUENCY_STEP_HZ,
    "kr_center_rad_per_m": KR_CENTER,
    "delta_kr_rad_per_m": DELTA_KR,
    "kx_center_rad_per_m": AZIMUTH_WAVENUMBER_CENTER,
    "delta_kx_rad_per_m": AZIMUTH_WAVENUMBER_STEP,
    "reference_range_m": REFERENCE_RANGE_M,
    "mixed_derivative_m": KAPPA_XR,
    "mixed_scale_per_m": MIXED_SCALE,
    "f_star_rad": F_STAR,
    "f_star_over_cell_width": F_STAR / CELL_WIDTH,
    "bilinear_coefficient": bilinear_coefficient,
    "bilinear_r2": bilinear_r2,
    "maximum_aligned_residual_rad": maximum_aligned_residual,
}

print(pd.Series(GEOMETRY).to_string(float_format=lambda value: f"{value:.6g}"))
```

## Phase-history generator

After removal of the separable phases known from the geometry, the complex patch for scatterers $j$ is modeled as

$$
s_{ad}=\sum_j \alpha_j\, e^{-i(u_j a+v_j d)}\, e^{i f_j H^{\rm exact}_{ad}}+\varepsilon_{ad},
\qquad
f_j=R_j\,\kappa_{xr}\,\Delta k_x\Delta k_r .
$$

The labeled target lies near $R_0$, with a small range jitter, and its azimuth cell is the label; its azimuth wavenumber $u_j$ is offset continuously within the cell. Off-range clutter scatterers have $R_j\in[0.45,1.55]\,R_0$ and are therefore not focused by the target query. The nuisance phases $v_j$, the reflectivities $\alpha_j$, the clutter, the complex Gaussian noise $\varepsilon_{ad}$ and a fixed azimuth-channel phase response are independent of the label. The array is a small unfocused spectral phase-history patch, the domain in which the geometry-dependent Fourier phases enter $\omega$-$k$ focusing, rather than a raw time-domain SAR echo.

The shared trainable readout can represent the inverse of the channel phase exactly. It is included so that the aligned task still requires learning instead of being solved by the fixed Fourier measurement alone. Each patch is normalized and amplitude-encoded into the six-qubit state.

The next cell builds the fixed channel phase from the binary register indices and defines reproducible seeds and exactly balanced labels.

```python
H_EXACT_JAX = jnp.asarray(H_EXACT)
H_COMPACT_JAX = jnp.asarray(H_COMPACT)
INDEX_JAX = jnp.asarray(INDEX)
_sensor_bits = (
    (np.arange(Q)[:, None] >> np.arange(N - 1, -1, -1)[None, :]) & 1
)
_sensor_z = 1.0 - 2.0 * _sensor_bits
SENSOR_PHASE = np.exp(-0.5j * (_sensor_z @ SENSOR_Z_PHASES))
SENSOR_PHASE_JAX = jnp.asarray(SENSOR_PHASE)


def seed_from(*parts):
    sequence = np.random.SeedSequence([BASE_SEED, *[int(p) for p in parts]])
    return int(sequence.generate_state(1, dtype=np.uint32)[0])


def balanced_labels(key, batch_size):
    if batch_size % Q:
        raise ValueError(f"batch_size={batch_size} must be divisible by Q={Q}")
    # Every consecutive block contains each class once. Consequently all
    # declared nested prefixes remain exactly balanced, not only the full pool.
    scores = jax.random.uniform(key, (batch_size // Q, Q))
    return jnp.argsort(scores, axis=1).reshape(-1)
```

`scenes` draws a batch of labeled, normalized patches at target coupling `f_geom`. The clutter, noise and jitter levels default to the configuration and can be changed per call.

```python
def scenes(
    key,
    batch_size,
    f_geom,
    *,
    snr_db=SNR_DB,
    n_clutter=N_CLUTTER,
    clutter_rms=CLUTTER_RMS,
    azimuth_jitter=AZIMUTH_SUBCELL_JITTER,
    target_range_jitter=TARGET_RANGE_JITTER,
):
    """Generate balanced target-cell labels and unfocused complex histories."""
    keys = jax.random.split(key, 13)
    labels = balanced_labels(keys[0], batch_size)

    u_center = TWO_PI * labels / Q
    u_target = jnp.mod(
        u_center
        + azimuth_jitter
        * CELL_WIDTH
        * jax.random.uniform(keys[1], (batch_size,), minval=-1.0, maxval=1.0),
        TWO_PI,
    )
    v_target = jax.random.uniform(keys[2], (batch_size,), minval=0.0, maxval=TWO_PI)
    f_target = f_geom * (
        1.0 + target_range_jitter * jax.random.normal(keys[3], (batch_size,))
    )
    target_reflectivity = jnp.exp(
        1j * jax.random.uniform(keys[4], (batch_size,), minval=0.0, maxval=TWO_PI)
    )

    target = target_reflectivity[:, None, None] * jnp.exp(
        -1j
        * (
            u_target[:, None, None] * INDEX_JAX[None, :, None]
            + v_target[:, None, None] * INDEX_JAX[None, None, :]
        )
        + 1j * f_target[:, None, None] * H_EXACT_JAX[None, :, :]
    )

    if n_clutter:
        u_clutter = jax.random.uniform(
            keys[5], (batch_size, n_clutter), minval=0.0, maxval=TWO_PI
        )
        v_clutter = jax.random.uniform(
            keys[6], (batch_size, n_clutter), minval=0.0, maxval=TWO_PI
        )
        range_ratio = jax.random.uniform(
            keys[7],
            (batch_size, n_clutter),
            minval=CLUTTER_RANGE_RATIO[0],
            maxval=CLUTTER_RANGE_RATIO[1],
        )
        clutter_reflectivity = (
            clutter_rms
            / np.sqrt(2 * n_clutter)
            * (
                jax.random.normal(keys[8], (batch_size, n_clutter))
                + 1j * jax.random.normal(keys[9], (batch_size, n_clutter))
            )
        )
        clutter_phase = (
            -1j
            * (
                u_clutter[:, :, None, None] * INDEX_JAX[None, None, :, None]
                + v_clutter[:, :, None, None] * INDEX_JAX[None, None, None, :]
            )
            + 1j
            * (f_geom * range_ratio)[:, :, None, None]
            * H_EXACT_JAX[None, None, :, :]
        )
        clutter = jnp.sum(
            clutter_reflectivity[:, :, None, None] * jnp.exp(clutter_phase),
            axis=1,
        )
    else:
        clutter = jnp.zeros_like(target)

    signal = target + clutter
    rms = jnp.sqrt(jnp.mean(jnp.abs(signal) ** 2, axis=(1, 2), keepdims=True))
    sigma = rms * 10 ** (-snr_db / 20) / np.sqrt(2)
    noise = sigma * (
        jax.random.normal(keys[10], signal.shape)
        + 1j * jax.random.normal(keys[11], signal.shape)
    )
    signal = (signal + noise) * SENSOR_PHASE_JAX[None, :, None]
    signal = signal / jnp.linalg.norm(signal.reshape(batch_size, -1), axis=1)[:, None, None]
    return signal, labels
```

Because $E^{(f)}$ is diagonal, we apply it elementwise to the patch and then trace out the range register to obtain the reduced azimuth state $\rho$. Its purity measures how well the query refocuses the target.

```python
def rho_azimuth(signal, f, *, exact_query=False):
    """Apply the fixed diagonal query and trace out the range register."""
    generator = H_EXACT_JAX if exact_query else H_COMPACT_JAX
    queried = signal * jnp.exp(-1j * f * generator)[None, :, :]
    # Contract d, the second register. Contracting a here would be a silent bug.
    return jnp.einsum("bad,bAd->baA", queried, jnp.conj(queried))


def purity(rho):
    trace = jnp.real(jnp.trace(rho, axis1=1, axis2=2))
    return jnp.real(jnp.einsum("bij,bji->b", rho, rho)) / trace**2


print("scene model loaded")
```

## Capacity-matched classifier

The six-qubit amplitude-encoded state is queried with

$$
E^{(f)}=\exp\!\left[-if\,\widetilde a\widetilde d\right].
$$

Since centered binary indices satisfy $\widetilde a=-\frac12\sum_i 2^{n-1-i}Z_i$, and similarly for $\widetilde d$, this gate is a product of $n^2$ commuting cross-register $R_{ZZ}$ rotations. The range register is then traced out. On the remaining three azimuth qubits all arms use the same 12-parameter residual circuit, followed by a fixed inverse discrete Fourier readout and a computational-basis measurement.

The Fourier readout is common to every arm and defines the azimuth cells. The trainable residual circuit can adapt to clutter and model mismatch. At zero parameters it is the identity, so the aligned model contains the matched Fourier readout without relying on a favorable initialization. No trainable classical head is added.

The next cell builds the residual circuit from $R_Y$ and $R_Z$ rotations and a CZ ring on the azimuth qubits, two layers deep, and the fixed Fourier readout.

```python
_I2 = jnp.eye(2, dtype=jnp.complex128)
_CZ = jnp.diag(jnp.array([1.0, 1.0, 1.0, -1.0], dtype=jnp.complex128))


def kron_all(matrices):
    result = jnp.eye(1, dtype=jnp.complex128)
    for matrix in matrices:
        result = jnp.kron(result, matrix)
    return result


def single_qubit_gate(angle, kind):
    if kind == "ry":
        c, s = jnp.cos(angle / 2), jnp.sin(angle / 2)
        return jnp.array([[c, -s], [s, c]], dtype=jnp.complex128)
    return jnp.diag(
        jnp.array([jnp.exp(-0.5j * angle), jnp.exp(0.5j * angle)])
    )


def cz_ring_matrix():
    diagonal = np.ones(Q)
    for state in range(Q):
        bits = [(state >> (N - 1 - q)) & 1 for q in range(N)]
        phase = 1
        for q in range(N):
            if bits[q] and bits[(q + 1) % N]:
                phase *= -1
        diagonal[state] = phase
    return jnp.diag(jnp.asarray(diagonal, dtype=jnp.complex128))


CZ_RING = cz_ring_matrix()
FOURIER_READOUT = jnp.asarray(
    np.exp(2j * np.pi * np.outer(INDEX, INDEX) / Q) / np.sqrt(Q)
)


def residual_unitary(parameters):
    parameters = jnp.asarray(parameters).reshape(READOUT_LAYERS, N, 2)
    unitary = jnp.eye(Q, dtype=jnp.complex128)
    for layer in range(READOUT_LAYERS):
        for qubit in range(N):
            for gate_index, kind in enumerate(("ry", "rz")):
                matrices = [_I2] * N
                matrices[qubit] = single_qubit_gate(
                    parameters[layer, qubit, gate_index], kind
                )
                unitary = kron_all(matrices) @ unitary
        unitary = CZ_RING @ unitary
    return unitary


def readout_unitary(parameters):
    return FOURIER_READOUT @ residual_unitary(parameters)
```

Class probabilities are the computational-basis populations after the readout. We train with the cross-entropy and report balanced accuracy.

```python
def probabilities(parameters, rho):
    unitary = readout_unitary(parameters)
    probabilities_ = jnp.real(
        jnp.einsum("ja,baA,jA->bj", unitary, rho, jnp.conj(unitary))
    )
    probabilities_ = jnp.clip(probabilities_, 1e-12, None)
    return probabilities_ / probabilities_.sum(axis=1, keepdims=True)


def cross_entropy(parameters, rho, labels):
    p = probabilities(parameters, rho)
    return -jnp.mean(jnp.log(p[jnp.arange(len(labels)), labels]))


def balanced_accuracy(labels, predictions):
    labels = np.asarray(labels, dtype=int)
    predictions = np.asarray(predictions, dtype=int)
    return float(
        np.mean(
            [np.mean(predictions[labels == cls] == cls) for cls in range(Q)]
        )
    )
```

Training uses Adam with gradient-norm clipping and a cosine learning-rate schedule, vectorized over restarts and compiled with JAX. Each restart keeps its lowest-loss parameters.

```python
def make_adam_runner():
    loss_and_gradient = jax.vmap(
        jax.value_and_grad(cross_entropy),
        in_axes=(0, None, None),
    )
    loss_batch = jax.vmap(cross_entropy, in_axes=(0, None, None))

    @jax.jit
    def run(initial_parameters, rho, labels):
        parameters = jnp.asarray(initial_parameters)
        first = jnp.zeros_like(parameters)
        second = jnp.zeros_like(parameters)
        best_parameters = parameters
        best_losses = loss_batch(parameters, rho, labels)

        def body(step, state):
            parameters, first, second, best_parameters, best_losses = state
            _, gradient = loss_and_gradient(parameters, rho, labels)
            gradient_norm = jnp.linalg.norm(
                gradient.reshape(gradient.shape[0], -1), axis=1
            )
            scale = jnp.minimum(1.0, 10.0 / (gradient_norm + 1e-12))
            gradient = gradient * scale.reshape((-1,) + (1,) * (gradient.ndim - 1))

            first = 0.9 * first + 0.1 * gradient
            second = 0.999 * second + 0.001 * gradient**2
            t = step + 1
            first_hat = first / (1.0 - 0.9**t)
            second_hat = second / (1.0 - 0.999**t)
            fraction = step / jnp.maximum(1, P["adam_steps"] - 1)
            rate = P["adam_lr"] * (
                0.1 + 0.9 * 0.5 * (1.0 + jnp.cos(jnp.pi * fraction))
            )
            parameters = parameters - rate * first_hat / (
                jnp.sqrt(second_hat) + 1e-8
            )
            new_losses = loss_batch(parameters, rho, labels)
            improved = new_losses < best_losses
            best_losses = jnp.where(improved, new_losses, best_losses)
            best_parameters = jnp.where(
                improved.reshape((-1,) + (1,) * (parameters.ndim - 1)),
                parameters,
                best_parameters,
            )
            return parameters, first, second, best_parameters, best_losses

        _, _, _, best_parameters, best_losses = jax.lax.fori_loop(
            0,
            P["adam_steps"],
            body,
            (parameters, first, second, best_parameters, best_losses),
        )
        return best_parameters, best_losses

    return run


ADAM_RUNNER = make_adam_runner()
```

Restarts use paired initializations shared by all arms, with the first restart at zero parameters, i.e. at the matched Fourier readout. `fit_readout` selects the best restart by training loss only and then evaluates it on the test set. `fixed_fourier_accuracy` evaluates the untrained readout, optionally with the known inverse channel phase, as a reference.

```python
def paired_initializations(seed):
    rng = np.random.default_rng(seed)
    shape = (P["n_restarts"], READOUT_LAYERS, N, 2)
    parameters = 0.15 * rng.normal(size=shape)
    parameters[0] = 0.0
    return parameters


def fit_readout(rho_train, labels_train, rho_test=None, labels_test=None, *, seed):
    initial = paired_initializations(seed)
    candidates, losses = ADAM_RUNNER(
        jnp.asarray(initial), jnp.asarray(rho_train), jnp.asarray(labels_train)
    )
    losses = np.asarray(losses)
    best_index = int(np.argmin(losses))
    best_parameters = np.asarray(candidates[best_index])
    result = {
        "train_loss": float(losses[best_index]),
        "best_restart": best_index,
    }
    if rho_test is not None:
        p = np.asarray(probabilities(best_parameters, rho_test))
        labels = np.asarray(labels_test, dtype=int)
        prediction = np.argmax(p, axis=1)
        result.update(
            test_accuracy=float(np.mean(prediction == labels)),
            test_balanced_accuracy=balanced_accuracy(labels, prediction),
            test_nll=float(-np.mean(np.log(p[np.arange(len(labels)), labels]))),
        )
    return result


def fixed_fourier_accuracy(rho, labels, *, calibrated=False):
    parameters = np.zeros((READOUT_LAYERS, N, 2))
    if calibrated:
        parameters[0, :, 1] = -SENSOR_Z_PHASES
    p = np.asarray(probabilities(parameters, rho))
    prediction = np.argmax(p, axis=1)
    return balanced_accuracy(labels, prediction)


print("classifier loaded")
```

## Validity checks

These checks run before any fitting. They verify the geometry derivative, the quality of the compact approximation, the $R_{ZZ}$ decomposition of the query, the register traced out by the classifier, the identity point of the residual circuit, class balance, and the expected coherence restoration for a noiseless single target.

We first compare $\kappa_{xr}$ with a central finite difference and check that $H^{\rm exact}$ has zero row and column means and is well described by $\widetilde a\widetilde d$.

```python
VALIDITY = {}

# analytic mixed derivative against a central finite difference
epsilon = 1e-3
ky_scalar = lambda kx, kr: np.sqrt(kr**2 - kx**2)
finite_mixed = (
    ky_scalar(AZIMUTH_WAVENUMBER_CENTER + epsilon, KR_CENTER + epsilon)
    - ky_scalar(AZIMUTH_WAVENUMBER_CENTER + epsilon, KR_CENTER - epsilon)
    - ky_scalar(AZIMUTH_WAVENUMBER_CENTER - epsilon, KR_CENTER + epsilon)
    + ky_scalar(AZIMUTH_WAVENUMBER_CENTER - epsilon, KR_CENTER - epsilon)
) / (4 * epsilon**2)
VALIDITY["mixed derivative relative error"] = float(
    abs(finite_mixed - KAPPA_XR) / abs(KAPPA_XR)
)
assert VALIDITY["mixed derivative relative error"] < 2e-5

# separable terms removed; compact approximation
VALIDITY["interaction row mean"] = float(np.max(np.abs(H_EXACT.mean(axis=1))))
VALIDITY["interaction column mean"] = float(np.max(np.abs(H_EXACT.mean(axis=0))))
VALIDITY["bilinear coefficient"] = bilinear_coefficient
VALIDITY["bilinear R2"] = bilinear_r2
VALIDITY["aligned residual max rad"] = maximum_aligned_residual
assert VALIDITY["interaction row mean"] < 1e-8
assert VALIDITY["interaction column mean"] < 1e-8
assert abs(bilinear_coefficient - 1.0) < 0.01
assert bilinear_r2 > 0.995
```

The bilinear query compiles exactly into $n^2$ cross-register $R_{ZZ}$ rotations. We also check that the residual circuit is the identity at zero parameters and that its first $R_Z$ layer reproduces the inverse channel phase exactly.

```python
weights = 2 ** np.arange(N - 1, -1, -1)
bits = ((np.arange(Q)[:, None] >> np.arange(N - 1, -1, -1)) & 1)
z = 1.0 - 2.0 * bits


def compiled_query_phase(f):
    exponent = np.zeros((Q, Q))
    for i, wi in enumerate(weights):
        for j, wj in enumerate(weights):
            theta = f * wi * wj / 2.0
            exponent += -0.5 * theta * np.outer(z[:, i], z[:, j])
    return np.exp(1j * exponent)


VALIDITY["RZZ decomposition"] = float(
    np.max(
        np.abs(
            compiled_query_phase(0.731)
            - np.exp(-1j * 0.731 * H_COMPACT)
        )
    )
)
VALIDITY["two-register RZZ gates"] = N * N
assert VALIDITY["RZZ decomposition"] < 1e-12

# the residual readout is the identity at zero parameters
zero_parameters = np.zeros((READOUT_LAYERS, N, 2))
VALIDITY["zero residual identity"] = float(
    np.max(np.abs(np.asarray(residual_unitary(zero_parameters)) - np.eye(Q)))
)
assert VALIDITY["zero residual identity"] < 1e-12

calibration_parameters = np.zeros((READOUT_LAYERS, N, 2))
calibration_parameters[0, :, 1] = -SENSOR_Z_PHASES
VALIDITY["sensor calibration exactness"] = float(
    np.max(
        np.abs(
            np.asarray(residual_unitary(calibration_parameters))
            - np.diag(np.conj(SENSOR_PHASE))
        )
    )
)
assert VALIDITY["sensor calibration exactness"] < 1e-12
```

On a batch of scenes the reduced states must be valid density matrices, which also confirms that the range register is the one traced out. Labels must be exactly balanced for every declared sample size.

```python
check_signal, check_labels = scenes(
    jax.random.PRNGKey(seed_from(10)), 8 * Q, F_STAR
)
check_rho = np.asarray(rho_azimuth(check_signal, F_STAR))
VALIDITY["trace error"] = float(
    np.max(np.abs(np.trace(check_rho, axis1=1, axis2=2) - 1.0))
)
VALIDITY["Hermiticity error"] = float(
    np.max(np.abs(check_rho - np.conj(np.swapaxes(check_rho, 1, 2))))
)
VALIDITY["minimum eigenvalue"] = float(np.linalg.eigvalsh(check_rho).min())
assert VALIDITY["trace error"] < 1e-11
assert VALIDITY["Hermiticity error"] < 1e-11
assert VALIDITY["minimum eigenvalue"] > -1e-10

# balanced labels for all declared sample sizes
declared_sizes = set(
    [P["n_train_scan"], P["n_test"], *P["efficiency_ns"]]
)
for size in declared_sizes:
    _, labels = scenes(jax.random.PRNGKey(seed_from(11, size)), size, F_STAR)
    counts = np.bincount(np.asarray(labels), minlength=Q)
    assert counts.max() == counts.min()
VALIDITY["balanced declared sizes"] = True
```

For a noiseless single target, the exact query restores a pure azimuth state, the compact query a nearly pure one, and no query leaves it close to maximally mixed. Finally, the calibrated untrained readout confirms that the noisy scene distribution contains the intended contrast between the arms before any learned result is produced.

```python
u0, v0 = 1.231, 2.137
clean = (
    np.exp(-1j * (u0 * INDEX[:, None] + v0 * INDEX[None, :]))
    * np.exp(1j * F_STAR * H_EXACT)
)[None] / Q
purities = {}
for name, f, exact in [
    ("exact", F_STAR, True),
    ("compact", F_STAR, False),
    ("wrong", F_WRONG, False),
    ("none", 0.0, False),
]:
    purities[name] = float(
        purity(rho_azimuth(jnp.asarray(clean), f, exact_query=exact))[0]
    )
VALIDITY["clean purity exact"] = purities["exact"]
VALIDITY["clean purity compact"] = purities["compact"]
VALIDITY["clean purity wrong"] = purities["wrong"]
VALIDITY["clean purity none"] = purities["none"]
assert abs(purities["exact"] - 1.0) < 1e-12
assert purities["compact"] > 0.98
assert purities["none"] < 0.2

# known inverse sensor phase, used only to audit the generator
probe_signal, probe_labels = scenes(
    jax.random.PRNGKey(seed_from(12)), 800, F_STAR
)
oracle_accuracy = {}
for name, f in ARMS.items():
    oracle_accuracy[name] = fixed_fourier_accuracy(
        rho_azimuth(probe_signal, f), probe_labels, calibrated=True
    )
VALIDITY["probe accuracy aligned"] = oracle_accuracy["aligned"]
VALIDITY["probe accuracy wrong"] = oracle_accuracy["wrong"]
VALIDITY["probe accuracy none"] = oracle_accuracy["none"]
assert oracle_accuracy["aligned"] > oracle_accuracy["wrong"] + 0.25
assert oracle_accuracy["aligned"] > 0.60
assert abs(oracle_accuracy["none"] - 1.0 / Q) < 0.08

for key, value in VALIDITY.items():
    if isinstance(value, bool):
        print(f"{key:38s} {value}")
    else:
        print(f"{key:38s} {value:.6g}")
print("all validity gates passed")
```

## Experiments

The figure rests on three predeclared experiments.

1. **Query scan.** At a fixed, sufficiently large training set, we retrain the classifier independently for each $f/f^\star_{\rm geom}\in[0,2]$.
2. **Geometry tracking.** We repeat the training-loss scan for several reference ranges. The empirical minimizer $\widehat f$ uses training loss only and is compared with $f^\star_{\rm geom}$ computed from the geometry.
3. **Sample efficiency.** We compare the aligned, mismatched and no-query arms on nested training subsets.

Within every data draw, all query values use identical scenes and paired initializations. Parameters are retrained from scratch, with no continuation across query values or sample sizes, and the best optimizer restart is selected by training loss. Test data are never used to choose $f$, a restart or an optimization budget.

The full configuration is hashed into `RUN_ID`, which names the checkpoint CSV and the stored configuration, so that incompatible runs are never combined.

```python
RUN_SPEC = {
    "profile": PROFILE,
    "register_qubits": N,
    "geometry": GEOMETRY,
    "snr_db": SNR_DB,
    "n_clutter": N_CLUTTER,
    "clutter_rms": CLUTTER_RMS,
    "azimuth_subcell_jitter": AZIMUTH_SUBCELL_JITTER,
    "target_range_jitter": TARGET_RANGE_JITTER,
    "sensor_z_phases": SENSOR_Z_PHASES.tolist(),
    "clutter_range_ratio": list(CLUTTER_RANGE_RATIO),
    "readout_layers": READOUT_LAYERS,
    "profile_settings": P,
    "global_scales": GLOBAL_SCALES.tolist(),
    "recovery_ranges_m": RECOVERY_RANGES.tolist(),
    "recovery_f_grid": RECOVERY_F_GRID.tolist(),
    "base_seed": BASE_SEED,
}
RUN_ID = hashlib.sha256(
    json.dumps(RUN_SPEC, sort_keys=True).encode("utf-8")
).hexdigest()[:12]
RAW_PATH = RESULTS_DIR / f"sar_geometry_{PROFILE}_{RUN_ID}.csv"
CONFIG_PATH = RESULTS_DIR / f"sar_geometry_{PROFILE}_{RUN_ID}_config.json"
CONFIG_PATH.write_text(json.dumps(RUN_SPEC, indent=2))
```

Each completed fit is appended to the CSV as one row, and completed task IDs are skipped on rerun.

```python
RAW_COLUMNS = [
    "run_id",
    "task_id",
    "experiment",
    "draw",
    "range_m",
    "f_geom",
    "f",
    "f_over_star",
    "arm",
    "n_train",
    "train_loss",
    "best_restart",
    "test_accuracy",
    "test_balanced_accuracy",
    "test_nll",
    "compact_qft_accuracy",
]


def completed_tasks():
    if not RAW_PATH.exists():
        return set()
    table = pd.read_csv(RAW_PATH)
    return set(table["task_id"].astype(str))


def append_result(row):
    complete_row = {column: row.get(column, np.nan) for column in RAW_COLUMNS}
    pd.DataFrame([complete_row], columns=RAW_COLUMNS).to_csv(
        RAW_PATH,
        mode="a",
        header=not RAW_PATH.exists(),
        index=False,
    )


def add_fit(row, result):
    append_result({**row, **result})
```

`run_all_experiments` loops over the three experiments. The efficiency experiment uses nested prefixes of one training pool and also records the calibrated fixed Fourier references on the test scenes.

```python
def run_all_experiments():
    completed = completed_tasks()
    t0 = time.time()

    # query scan
    for draw in range(P["n_draws"]):
        train_signal, train_labels = scenes(
            jax.random.PRNGKey(seed_from(100, draw)),
            P["n_train_scan"],
            F_STAR,
        )
        test_signal, test_labels = scenes(
            jax.random.PRNGKey(seed_from(101, draw)),
            P["n_test"],
            F_STAR,
        )
        paired_seed = seed_from(102, draw)
        for index, (scale, f) in enumerate(zip(GLOBAL_SCALES, GLOBAL_F)):
            task_id = f"scan-d{draw:03d}-f{index:03d}"
            if task_id in completed:
                continue
            result = fit_readout(
                rho_azimuth(train_signal, f),
                train_labels,
                rho_azimuth(test_signal, f),
                test_labels,
                seed=paired_seed,
            )
            add_fit(
                {
                    "run_id": RUN_ID,
                    "task_id": task_id,
                    "experiment": "scan",
                    "draw": draw,
                    "range_m": REFERENCE_RANGE_M,
                    "f_geom": F_STAR,
                    "f": float(f),
                    "f_over_star": float(scale),
                    "arm": "scan",
                    "n_train": P["n_train_scan"],
                },
                result,
            )
            completed.add(task_id)
            print(
                f"scan draw={draw:02d} f/f*={scale:.3f} "
                f"acc={result['test_balanced_accuracy']:.3f}",
                flush=True,
            )

    # geometry tracking
    n_before = len(completed)
    for geometry_index, (range_m, f_geom) in enumerate(
        zip(RECOVERY_RANGES, RECOVERY_F_STAR)
    ):
        for draw in range(P["recovery_draws"]):
            train_signal, train_labels = scenes(
                jax.random.PRNGKey(seed_from(200, geometry_index, draw)),
                P["n_train_scan"],
                float(f_geom),
            )
            paired_seed = seed_from(201, geometry_index, draw)
            for f_index, f in enumerate(RECOVERY_F_GRID):
                task_id = (
                    f"recovery-g{geometry_index:02d}-d{draw:03d}-f{f_index:03d}"
                )
                if task_id in completed:
                    continue
                result = fit_readout(
                    rho_azimuth(train_signal, f),
                    train_labels,
                    seed=paired_seed,
                )
                add_fit(
                    {
                        "run_id": RUN_ID,
                        "task_id": task_id,
                        "experiment": "recovery",
                        "draw": draw,
                        "range_m": float(range_m),
                        "f_geom": float(f_geom),
                        "f": float(f),
                        "f_over_star": float(f / f_geom),
                        "arm": "scan",
                        "n_train": P["n_train_scan"],
                    },
                    result,
                )
                completed.add(task_id)
            if len(completed) > n_before:
                print(
                    f"recovery range={range_m:.0f} m draw={draw:02d} complete",
                    flush=True,
                )
            n_before = len(completed)

    # sample efficiency
    n_max = max(P["efficiency_ns"])
    n_before = len(completed)
    for draw in range(P["n_draws"]):
        pool_signal, pool_labels = scenes(
            jax.random.PRNGKey(seed_from(300, draw)), n_max, F_STAR
        )
        test_signal, test_labels = scenes(
            jax.random.PRNGKey(seed_from(301, draw)), P["n_test"], F_STAR
        )

        # Fixed Fourier references are properties of the same test scenes.
        reference_task = f"reference-d{draw:03d}"
        if reference_task not in completed:
            exact_accuracy = fixed_fourier_accuracy(
                rho_azimuth(test_signal, F_STAR, exact_query=True),
                test_labels,
                calibrated=True,
            )
            compact_accuracy = fixed_fourier_accuracy(
                rho_azimuth(test_signal, F_STAR),
                test_labels,
                calibrated=True,
            )
            append_result(
                {
                    "run_id": RUN_ID,
                    "task_id": reference_task,
                    "experiment": "reference",
                    "draw": draw,
                    "range_m": REFERENCE_RANGE_M,
                    "f_geom": F_STAR,
                    "f": F_STAR,
                    "f_over_star": 1.0,
                    "arm": "exact_qft",
                    "n_train": 0,
                    "train_loss": np.nan,
                    "best_restart": np.nan,
                    "test_accuracy": exact_accuracy,
                    "test_balanced_accuracy": exact_accuracy,
                    "test_nll": np.nan,
                    "compact_qft_accuracy": compact_accuracy,
                }
            )
            completed.add(reference_task)

        for n_train in P["efficiency_ns"]:
            paired_seed = seed_from(302, draw, n_train)
            for arm, f in ARMS.items():
                task_id = f"efficiency-d{draw:03d}-n{n_train:05d}-{arm}"
                if task_id in completed:
                    continue
                result = fit_readout(
                    rho_azimuth(pool_signal[:n_train], f),
                    pool_labels[:n_train],
                    rho_azimuth(test_signal, f),
                    test_labels,
                    seed=paired_seed,
                )
                add_fit(
                    {
                        "run_id": RUN_ID,
                        "task_id": task_id,
                        "experiment": "efficiency",
                        "draw": draw,
                        "range_m": REFERENCE_RANGE_M,
                        "f_geom": F_STAR,
                        "f": float(f),
                        "f_over_star": float(f / F_STAR),
                        "arm": arm,
                        "n_train": int(n_train),
                    },
                    result,
                )
                completed.add(task_id)
            if len(completed) > n_before:
                print(
                    f"efficiency draw={draw:02d} N={n_train:04d} complete",
                    flush=True,
                )
            n_before = len(completed)

    results = pd.read_csv(RAW_PATH)
    print(
        f"loaded {len(results)} completed rows from {RAW_PATH} | "
        f"this call: {time.time() - t0:.1f}s"
    )
    return results
```

## Aggregation

Each data draw is one independent experimental unit. Curves are paired within a draw, and 95% bootstrap intervals resample draws rather than treating query points, optimizer restarts or nested training subsets as independent.

The next cell defines the draw bootstrap and summarizes the query scan.

```python
PRIMARY_METRIC = "test_balanced_accuracy"


def bootstrap_matrix(values, *, seed):
    values = np.asarray(values, dtype=float)
    rng = np.random.default_rng(seed_from(500, seed))
    indices = rng.integers(
        0, values.shape[0], (P["bootstrap_samples"], values.shape[0])
    )
    draws = values[indices].mean(axis=1)
    return {
        "mean": values.mean(axis=0),
        "lower": np.quantile(draws, 0.025, axis=0),
        "upper": np.quantile(draws, 0.975, axis=0),
        "draws": draws,
    }


SCAN = RAW_RESULTS[RAW_RESULTS["experiment"] == "scan"].copy()
scan_pivot = SCAN.pivot(
    index="draw", columns="f_over_star", values=PRIMARY_METRIC
).sort_index(axis=1)
SCAN_X = scan_pivot.columns.to_numpy(dtype=float)
SCAN_SUMMARY = bootstrap_matrix(scan_pivot.to_numpy(), seed=1)
```

For geometry tracking, $\widehat f$ is obtained from a local quadratic interpolation around the minimum sampled training loss. The interpolation is used only when it is convex and its vertex lies between the neighboring grid points; otherwise the sampled minimizer is kept. We then average $\widehat f$ over draws for each reference range.

```python
def interpolated_argmin(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    index = int(np.argmin(y))
    if index == 0 or index == len(x) - 1:
        return float(x[index]), "grid_edge"
    x_local = x[index - 1 : index + 2]
    y_local = y[index - 1 : index + 2]
    a, b, _ = np.polyfit(x_local, y_local, 2)
    if a <= 0:
        return float(x[index]), "sampled"
    vertex = -b / (2 * a)
    if not (x_local[0] <= vertex <= x_local[-1]):
        return float(x[index]), "sampled"
    return float(vertex), "quadratic"


RECOVERY = RAW_RESULTS[RAW_RESULTS["experiment"] == "recovery"].copy()
recovery_rows = []
for (range_m, f_geom, draw), part in RECOVERY.groupby(
    ["range_m", "f_geom", "draw"]
):
    part = part.sort_values("f")
    f_hat, method = interpolated_argmin(part["f"], part["train_loss"])
    recovery_rows.append(
        {
            "range_m": range_m,
            "f_geom": f_geom,
            "draw": int(draw),
            "f_hat": f_hat,
            "method": method,
            "absolute_error": abs(f_hat - f_geom),
        }
    )
RECOVERY_OPTIMA = pd.DataFrame(recovery_rows)

recovery_summary_rows = []
for f_geom, part in RECOVERY_OPTIMA.groupby("f_geom"):
    values = part["f_hat"].to_numpy()[:, None]
    summary = bootstrap_matrix(values, seed=2 + len(recovery_summary_rows))
    recovery_summary_rows.append(
        {
            "f_geom": float(f_geom),
            "f_hat": float(summary["mean"][0]),
            "lower95": float(summary["lower"][0]),
            "upper95": float(summary["upper"][0]),
        }
    )
RECOVERY_SUMMARY = pd.DataFrame(recovery_summary_rows).sort_values("f_geom")
```

We summarize the sample-efficiency curves of each arm, the fixed Fourier references, and the paired differences between the aligned arm and each control at the largest training set.

```python
EFFICIENCY = RAW_RESULTS[RAW_RESULTS["experiment"] == "efficiency"].copy()
EFFICIENCY_SUMMARY = {}
for arm in ARMS:
    pivot = EFFICIENCY[EFFICIENCY["arm"] == arm].pivot(
        index="draw", columns="n_train", values=PRIMARY_METRIC
    ).sort_index(axis=1)
    EFFICIENCY_SUMMARY[arm] = {
        "n": pivot.columns.to_numpy(dtype=int),
        **bootstrap_matrix(pivot.to_numpy(), seed=20 + len(EFFICIENCY_SUMMARY)),
        "per_draw": pivot,
    }

REFERENCE = RAW_RESULTS[RAW_RESULTS["experiment"] == "reference"].copy()
EXACT_QFT_REFERENCE = float(REFERENCE[PRIMARY_METRIC].mean())
COMPACT_QFT_REFERENCE = float(REFERENCE["compact_qft_accuracy"].mean())

# paired effects at the largest N
N_LARGEST = max(P["efficiency_ns"])
largest = EFFICIENCY[EFFICIENCY["n_train"] == N_LARGEST].pivot(
    index="draw", columns="arm", values=PRIMARY_METRIC
)
PAIRED_EFFECTS = {}
for control in ("wrong", "none"):
    difference = (largest["aligned"] - largest[control]).to_numpy()
    summary = bootstrap_matrix(difference[:, None], seed=40 + len(PAIRED_EFFECTS))
    PAIRED_EFFECTS[control] = {
        "mean": float(summary["mean"][0]),
        "lower95": float(summary["lower"][0]),
        "upper95": float(summary["upper"][0]),
    }
```

The key numbers are the query-scan peak position, the RMS error of $\widehat f$ against $f^\star_{\rm geom}$, the calibrated Fourier references and the paired accuracy gains at the largest training set.

```python
SCAN_PEAK_SCALE = float(SCAN_X[np.argmax(SCAN_SUMMARY["mean"])])
RECOVERY_RMSE = float(
    np.sqrt(np.mean((RECOVERY_OPTIMA["f_hat"] - RECOVERY_OPTIMA["f_geom"]) ** 2))
)

KEY_RESULTS = {
    "scan_peak_f_over_fstar": SCAN_PEAK_SCALE,
    "recovery_rmse_rad": RECOVERY_RMSE,
    "exact_query_qft_reference": EXACT_QFT_REFERENCE,
    "compact_query_qft_reference": COMPACT_QFT_REFERENCE,
    "aligned_minus_wrong_at_largest_N": PAIRED_EFFECTS["wrong"],
    "aligned_minus_none_at_largest_N": PAIRED_EFFECTS["none"],
}
print(json.dumps(KEY_RESULTS, indent=2))

display(RECOVERY_SUMMARY)
display(
    pd.DataFrame(PAIRED_EFFECTS).T.rename_axis("control")
)
```

We first define the axis helpers and the color, label and marker of each arm.

```python
def panel_title(axis, letter, description):
    """Bold panel letter on the left, description on the right above the plot."""
    axis.set_title(letter, loc="left", pad=4, fontweight="bold")
    axis.set_title(description, loc="right", pad=4, fontsize=7.0, color="0.25")


def finish_axis(axis, x_minor=True, y_minor=True, grid_axis="both"):
    """Gridlines, minor ticks and spines applied consistently across panels."""
    axis.set_axisbelow(True)
    axis.minorticks_on()
    if not x_minor:
        axis.xaxis.set_minor_locator(mticker.NullLocator())
    if not y_minor:
        axis.yaxis.set_minor_locator(mticker.NullLocator())
    axis.tick_params(which="both", top=False, right=False)
    axis.grid(True, which="major", axis=grid_axis, linestyle="-", alpha=0.30)
    axis.grid(True, which="minor", axis=grid_axis, linestyle=":", alpha=0.15)
    for spine in axis.spines.values():
        spine.set_linewidth(0.8)


COLORS = {
    "aligned": "#E64B35",
    "wrong": "#4DBBD5",
    "none": "#7E6148",
    "scan": "#3C5488",
    "geometry": "#00A087",
    "reference": "#4D4D4D",
}
LABELS = {
    "aligned": r"aligned query $f=f^\star_{\rm geom}$",
    "wrong": r"mismatched $f=f^\star_{\rm geom}/2$",
    "none": r"no query $f=0$",
}
MARKERS = {"aligned": "o", "wrong": "s", "none": "^"}
```

We draw the three panels with a shared legend and a common accuracy range for panels (a) and (c), and save the figure.

```python
# double-column width
fig, axes = plt.subplots(1, 3, figsize=(7.15, 2.75))

# (a) query alignment
axis_a = axes[0]
axis_a.fill_between(
    SCAN_X,
    SCAN_SUMMARY["lower"],
    SCAN_SUMMARY["upper"],
    color=COLORS["scan"],
    alpha=0.18,
    lw=0,
)
axis_a.plot(
    SCAN_X,
    SCAN_SUMMARY["mean"],
    "-o",
    color=COLORS["scan"],
    ms=2.6,
    markevery=max(1, len(SCAN_X) // 12),
)
axis_a.axvline(1.0, color=COLORS["geometry"], ls="--", lw=0.9)

for scale, arm in [(0.0, "none"), (0.5, "wrong"), (1.0, "aligned")]:
    idx = int(np.argmin(np.abs(SCAN_X - scale)))
    axis_a.plot(
        SCAN_X[idx],
        SCAN_SUMMARY["mean"][idx],
        marker=MARKERS[arm],
        color=COLORS[arm],
        mec="white",
        mew=0.45,
        ms=4.4,
        ls="none",
        zorder=5,
    )

axis_a.set_xlim(0.0, 2.0)
axis_a.set_xticks([0.0, 0.5, 1.0, 1.5, 2.0])
axis_a.set_xlabel(r"Query scale $f/f^\star_{\rm geom}$")
axis_a.set_ylabel("Test balanced accuracy")
panel_title(axis_a, "(a)", "Query alignment")
finish_axis(axis_a, x_minor=True)

# (b) geometry tracking
axis_b = axes[1]
x = RECOVERY_SUMMARY["f_geom"].to_numpy()
y = RECOVERY_SUMMARY["f_hat"].to_numpy()
low = y - RECOVERY_SUMMARY["lower95"].to_numpy()
high = RECOVERY_SUMMARY["upper95"].to_numpy() - y
limit_low = min(x.min(), y.min()) - 0.06
limit_high = max(x.max(), y.max()) + 0.06

axis_b.plot(
    [limit_low, limit_high],
    [limit_low, limit_high],
    color=COLORS["reference"],
    lw=0.8,
    ls="--",
)
axis_b.errorbar(
    x,
    y,
    yerr=np.vstack([low, high]),
    fmt="o",
    color=COLORS["geometry"],
    ecolor=COLORS["geometry"],
    elinewidth=0.8,
    capsize=2,
    ms=4.0,
    mec="white",
    mew=0.45,
)

axis_b.set_xlim(limit_low, limit_high)
axis_b.set_ylim(limit_low, limit_high)
axis_b.set_xlabel(r"Geometry prediction $f^\star_{\rm geom}$ (rad)")
axis_b.set_ylabel(r"Training-only optimum $\widehat f$ (rad)")
panel_title(axis_b, "(b)", "Geometry tracking")
finish_axis(axis_b, x_minor=True)

# (c) sample efficiency
axis_c = axes[2]
for arm in ("aligned", "wrong", "none"):
    summary = EFFICIENCY_SUMMARY[arm]
    axis_c.fill_between(
        summary["n"],
        summary["lower"],
        summary["upper"],
        color=COLORS[arm],
        alpha=0.14,
        lw=0,
    )
    axis_c.plot(
        summary["n"],
        summary["mean"],
        marker=MARKERS[arm],
        color=COLORS[arm],
        ms=3.5,
    )

axis_c.axhline(
    1.0 / Q,
    color=COLORS["reference"],
    lw=0.7,
    ls=":",
)
axis_c.axhline(
    EXACT_QFT_REFERENCE,
    color=COLORS["geometry"],
    lw=0.8,
    ls="--",
)

axis_c.set_xscale("log", base=2)
axis_c.set_xticks(P["efficiency_ns"])
axis_c.set_xticklabels([str(val) for val in P["efficiency_ns"]])
axis_c.set_xlabel("Training scenes $N$")
axis_c.set_ylabel("Test balanced accuracy")
panel_title(axis_c, "(c)", "Sample efficiency")
finish_axis(axis_c, x_minor=False)

# common accuracy range for panels (a) and (c)
all_lower = np.concatenate(
    [
        SCAN_SUMMARY["lower"],
        *[EFFICIENCY_SUMMARY[arm]["lower"] for arm in ARMS],
        [1.0 / Q],
    ]
)
all_upper = np.concatenate(
    [
        SCAN_SUMMARY["upper"],
        *[EFFICIENCY_SUMMARY[arm]["upper"] for arm in ARMS],
        [EXACT_QFT_REFERENCE],
    ]
)
padding = max(0.025, 0.08 * (all_upper.max() - all_lower.min()))
y_limits = (
    max(0.0, all_lower.min() - padding),
    min(1.01, all_upper.max() + padding),
)
axes[0].set_ylim(*y_limits)
axes[2].set_ylim(*y_limits)

# shared legend above the panels
legend_handles = [
    # row 1: query arms
    Line2D([], [], color=COLORS["aligned"], marker=MARKERS["aligned"], ms=4.0, label=LABELS["aligned"]),
    Line2D([], [], color=COLORS["wrong"], marker=MARKERS["wrong"], ms=4.0, label=LABELS["wrong"]),
    Line2D([], [], color=COLORS["none"], marker=MARKERS["none"], ms=4.0, label=LABELS["none"]),
    # row 2: scan and references
    Line2D([], [], color=COLORS["scan"], ls="-", marker="o", ms=2.6, label="query scale scan"),
    Line2D([], [], color=COLORS["geometry"], ls="--", lw=0.8, label="exact-query Fourier ref."),
    Line2D([], [], color=COLORS["reference"], ls=":", lw=0.7, label=f"chance baseline $1/{Q}$"),
]

legend = fig.legend(
    handles=legend_handles,
    loc="upper center",
    bbox_to_anchor=(0.5, 0.998),
    ncol=3,
    frameon=True,
    framealpha=0.9,
    edgecolor="black",
    fancybox=False,
    handletextpad=0.4,
)
legend.get_frame().set_linewidth(0.6)

fig.subplots_adjust(
    left=0.075,
    right=0.985,
    bottom=0.150,
    top=0.760,
    wspace=0.34,
)

for suffix, options in ((".pdf", {}), (".png", {"dpi": 600})):
    fig.savefig(
        FIGURES_DIR / f"{FIGURE_STEM}{suffix}",
        pad_inches=0.05,
        bbox_inches="tight",
        **options,
    )
plt.show()
print(f"saved {FIGURES_DIR / FIGURE_STEM}.pdf/.png")
```

## Scope and references

This experiment isolates one local focusing mechanism. It does not model the transmitted pulse, antenna pattern, platform motion errors, radiometric calibration, speckle statistics of an extended scene, or the complete Stolt interpolation pipeline. The input is therefore an unfocused SAR-like spectral phase history, not operational raw SAR data.

The wavenumber-domain construction follows the standard range-migration relation $k_y=\sqrt{k_r^2-k_x^2}$, for which Stolt interpolation maps the sampled $(k_x,k_r)$ data to a Cartesian wavenumber grid:

- B. Hosseiny *et al.*, “Range Migration Algorithm in the Processing Chain of Signals of a Ground-Based SAR Sensor,” *ISPRS Archives* XLII-4/W18, 521–525 (2019), [doi:10.5194/isprs-archives-XLII-4-W18-521-2019](https://doi.org/10.5194/isprs-archives-XLII-4-W18-521-2019).
- A. Reigber *et al.*, “Extended wavenumber-domain synthetic aperture radar focusing with integrated motion compensation,” *IEE Proc. Radar Sonar Navig.* 153, 301–310 (2006), [doi:10.1049/ip-rsn:20045087](https://doi.org/10.1049/ip-rsn:20045087).
